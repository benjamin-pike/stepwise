```python
from shutil import copyfile

from paths import paths
```

```python
OUT = paths.results.accompanying
OUT.mkdir(parents = True, exist_ok = True)

inputs = paths.analysis.experiments
```

```python
# Index structure and neighbourhood characteristics
structure = inputs['index/structure']

for source in [
    structure.descriptives_csv,
    structure.index_structure_csv,
    structure.pca_variance_csv,
    structure.component_correlations_csv,
    structure.contextual_correlations_csv,
]:
    copyfile(source, OUT / source.name)
```

```python
# Associations, nonlinearity, and effect modification
primary = inputs['associations/primary']
nonlinearity = inputs['associations/dose_response']
interactions = inputs['associations/effect_modification']

for source in [
    primary.sequential_models_csv,
    nonlinearity.spline_tests_csv,
    nonlinearity.quintile_contrasts_csv,
    interactions.interaction_models_csv,
    primary.unadjusted_correlations_csv,
]:
    copyfile(source, OUT / source.name)
```

```python
# Spatial models, constrained null, and prediction
boroughs = inputs['spatial/borough_fixed_effects']
dependence = inputs['spatial/autocorrelation_and_error_models']
radial = inputs['spatial/centre_periphery_adjustment']
conditional_null = inputs['spatial/conditional_null']
cross_validation = inputs['prediction/spatial_cross_validation']

for source in [
    boroughs.borough_models_csv,
    boroughs.borough_residual_moran_csv,
    dependence.spatial_error_models_csv,
    radial.centre_periphery_models_csv,
    conditional_null.conditional_null_summary_csv,
    cross_validation.spatial_cross_validation_csv,
    dependence.global_moran_csv,
    conditional_null.null_calibration_csv,
]:
    copyfile(source, OUT / source.name)
```

```python
# Index, regression, walkshed, and QOF sensitivity
index_definition = inputs['robustness/index_definition']
diagnostics = inputs['robustness/regression_diagnostics']
walkshed = inputs['robustness/walkshed_duration']
qof = inputs['robustness/qof_allocation']

for source in [
    index_definition.index_sensitivity_csv,
    index_definition.rank_correlations_csv,
    diagnostics.hc3_models_csv,
    diagnostics.influential_observations_csv,
    diagnostics.vif_csv,
    walkshed.walkshed_rank_correlations_csv,
    walkshed.walkshed_models_csv,
    qof.qof_allocation_mc_summary_csv,
]:
    copyfile(source, OUT / source.name)
```
